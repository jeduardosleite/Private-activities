````python
x = 1

while x <= 10:
    print(x)
    x += 1
````

````python
media = 1

while media <= 2:
  nota_1 = float(
      input(f'Digite a primeira nota: ')
  )

  nota_2 = float(
      input('Digite a segunda nota: ')
  )

  media_notas = (nota_1 + nota_2) / 2

  print(f'\nA média das notas foi de {media_notas}.\n')
  media += 1
````

## **Lista**

Podemos criar uma lista com *[]* e adicionar elementos, separando-os por vírgula.

```python
lista = ['string', 9.0, True]
```

Pode ser uma `string` entre parênteses simples, um `float`, valor `booleano` etc.

<br>

### **Índices**

Maneira de acessar um determinado elemento dentro da lista.
- *Valores positivos* : acessamos do primeiro -> último
- *Valores negativos* : acessamos do último -> primeiro

````python
"""
  Criei uma lista com o meu nome, notas e um valor booleano para "aprovado", sendo:
  - True = Aprovado
  - False = Reprovado
"""

lista = ['José Eduardo', 9.0, 9.5, 8.9, True]
print(lista)
````

````python
"""
  Acessando os índices positivos e negativos.
"""

primeiro_elemento = lista[0]
ultimo_elemento = lista[-1]

print(f'Usando o índice positivo, temos o primeiro elemento: {primeiro_elemento}')
print(f'Usando o índice negativo, temos o último elemento: {ultimo_elemento}')
````

### **Partição**

Utilizado para segregar uma lista, dividir os elementos em partes.

````python
print(lista)

elementos_1_3 = lista[1:3]
elemento_2_ultimo = lista[2:]

print('\n')
print(elementos_1_3)
print(elemento_2_ultimo)
````

5) Faça um programa que, ao inserir um número qualquer, cria uma lista contendo todos os números primos entre 1 e o número digitado.

````python
numero = int(input('Digite um número inteiro: '))
lista_primos = []

# laço que vai rodar todos os valores abaixo do número digitado
for num in range(2, numero):
  primo = True

  # Testamos se todos os números abaixo do especificado no primeio laço podem gerar uma divisão exata
  for divisiveis in range(2, num):
    if num % divisiveis == 0:
      primo = False  # Caso não seja divisível, o número não é primo
      break          # finalizamos o laço

  """
    A condição se torna o resultado booleano:
      False = náo é primo
      True = é primo
  """

  if primo:
    lista_primos.append(num)  # Caso seja primo, adicionamos ele à lista

print(f'\nLista de números primos: {lista_primos}')
print(f'\nAo total, temos {len(lista_primos)} números primos.')
````

6) Escreva um programa que peça uma data informando o dia, mês e ano e determine se ela é válida para uma análise.

````python
dia = int(input('Digite o dia: '))
mes = int(input('Digite o mês: '))
ano = int(input('Digite o ano: '))

# Meses com 30 e 31 dias
meses_30 = [4, 6, 9, 11]
meses_31 = [1, 3, 5, 7, 8, 10, 12]

# Verifica se o ano é bissexto
bissexto = ano % 4 == 0 and (ano % 400 == 0 or ano % 100 != 0)

# Define a quantidade de dias do mês
if mes == 2:
  max_dias = 29 if bissexto else 28

elif mes in meses_30:
  max_dias = 30

elif mes in meses_31:
  max_dias = 31

else:
  max_dias = 0

# Valida a data
if ano >= 1 and 1 <= dia <= max_dias:
  print('Data válida.')
else:
  print('Data inválida')
````

7) Para um estudo envolvendo o nível de multiplicação de bactérias em uma colônia, foi coletado o número de bactérias por dia (em milhares) e pode ser observado a seguir: [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]. Tendo esses valores, faça um código que gere uma lista contendo o percentual de crescimento de bactérias por dia, comparando o número de bactérias em cada dia com o número de bactérias do dia anterior. Dica: para calcular o percentual de crescimento usamos a seguinte equação:
```
100 * (amostra_atual - amostra_passada) / (amostra_passada).
```


````python
bacterias = [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]
crescimento = []

for i in range(1, len(bacterias)):
  porcentagem = round(100 * (bacterias[i] - bacterias[i - 1]) / (bacterias[i-1]), 2)
  crescimento.append(f'{porcentagem}%')

print(f'Porcentagens de crescimento: \n{crescimento}')
````

````python
respostas = []
gabarito = ['D', 'A', 'C', 'B', 'A', 'D', 'C', 'C', 'A', 'B']
nota = 0

# Recebemos as respostas do aluno
for i in range(0, 10):
  while True:
    resposta = input(f'Insira a resposta da questão {i +1}: ').upper()

    if resposta in ['A', 'B', 'C', 'D']:
      respostas.append(resposta)
      break
    else:
      print('Resposta inválida. Digite A, B, C ou D.')


# Verifica se as respostas são iguais e adicionamos à nota
for i in range(0, 10):
  if respostas[i] == gabarito[i]:
    nota += 1

print(f'A nota final foi {nota}.')
````

10) Um instituto de meteorologia deseja fazer um estudo de temperatura média de cada mês do ano. Para isso, você precisa fazer um código que colete e armazene essas temperaturas médias em uma lista. Depois, calcule a média anual das temperaturas e mostre todas as temperaturas acima da média anual e em que mês elas ocorreram, mostrando os meses por extenso (Janeiro, Fevereiro, etc.).

````python
temperaturas = []
media_anual = 0
meses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']

for i in range (0, len(meses)):
  while True:
    try:
      temperatura = float(input(f'Digite a temperatura média de {meses[i]}: '))
      temperaturas.append(temperatura)
      break
    except ValueError:
      print('Digite um valor válido.')

media_anual = sum(temperaturas) / len(temperaturas)
print(f'\nTemperatura média anual é de {media_anual:.2f}°C.\n')

# Resultado final
print(f'Temperaturas acima da média em: ')
for i in range(0, 12):
  if temperaturas[i] > media_anual:
    print(f'{meses[i]}: {temperaturas[i]:.2f}°C')
````

13) As pessoas colaboradoras de um setor da empresa que você trabalha vão receber um abono correspondente a 10% do salário devido ao ótimo desempenho do time. O setor financeiro solicitou sua ajuda para a verificação das consequências financeiras que esse abono irá gerar nos recursos. Assim, foi encaminhada para você uma lista com os salários que receberão o abono: [1172, 1644, 2617, 5130, 5532, 6341, 6650, 7238, 7685, 7782, 7903]. O abono de cada colaborador(a) não pode ser inferior a 200. Em código, transforme cada um dos salários em chaves de um dicionário e o abono de cada salário no elemento. Depois, informe o total de gastos com o abono, quantos(as) colaboradores(as) receberam o abono mínimo e qual o maior valor de abono fornecido.



````python
salarios = [1172, 1644, 2617, 5130, 5532, 6341, 6650, 7238, 7685, 7782, 7903]
abonos = {}

for salario in salarios:
  abono = round(salario * 0.1, 2)
  if abono < 200:
    abono = 200
  abonos[salario] = abono

total_gasto = sum(abonos.values())
qtd_abono_min = sum(1 for abono in abonos.values() if abono == 200)
maior_abono = max(abonos.values())

print(f'Total gasto com abonos: R$ {total_gasto}.')
print(f'Quantidade de colaboradores com abono mínimo: {qtd_abono_min}.')
print(f'Maior abono foi de R$ {maior_abono}.')
````

14) Uma equipe de cientistas de dados está estudando a diversidade biológica em uma floresta. A equipe fez a coleta de informações sobre o número de espécies de plantas e animais em cada área dessa floresta e armazenou essas informações em um dicionário. Nele, a chave descreve a área dos dados e os valores nas listas correspondem às espécies de plantas e animais nas áreas. Escreva um código para calcular a média de espécies por área e identificar a área com a maior diversidade biológica. Dica: use as funções built-in sum() e len().

```
{'Área Norte': [2819, 7236],
 'Área Leste': [1440, 9492],
 'Área Sul': [5969, 7496],
 'Área Oeste': [14446, 49688],
 'Área Centro': [22558, 45148]}
```

````python
especies = {
    'Área Norte': [2819, 7236],
    'Área Leste': [1440, 9492],
    'Área Sul': [5969, 7496],
    'Área Oeste': [14446, 49688],
    'Área Centro': [22558, 45148]
    }

media_especies = {}

for area, especies_area in especies.items():
  media = sum(especies_area) / len(especies_area)
  media_especies[area] = media

area_mais_diversa = max(media_especies, key=media_especies.get)

print(f'Média de espécies por área: {media_especies}')
print(f'Área com maior diversidade biológica: {area_mais_diversa}')
````

````python
setores = {
    'Setor A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
    'Setor B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
    'Setor C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
    'Setor D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]
    }

media_idades = {}

for setor, idades in setores.items():
  media = sum(idades) / len(idades)
  media_idades[setor] = round(media, 0)

print(f'A média geral ficou em {media:.0f} anos.')
print(f'Por setor, temos as médias {media_idades}')
````
